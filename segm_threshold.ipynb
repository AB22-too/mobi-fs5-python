**[Task 1.2 (d).]()** Binarize the image by applying an intensity threshold:

```python

plt.imshow(cells, cmap = 'gray')
plt.imsave('cells_gray.png', cells, cmap = 'gray')
gray = plt.imread('cells_gray.png')
```

```python
plt.hist(gray.ravel(), bins = 50)
```

```python
plt.imshow(gray)
plt.figure(figsize = (15,8))
```

```python
T = 0.1
binary = cells >= T
plt.imshow(binary, cmap= 'gray')
plt.show()
```

**[Task 1.2 (e).]()** Load the ground truth:

```python
ground = plt.imread('data/NIH3T3/gt/0.png')
plt.imshow(ground, cmap = 'gray')
```

Compute and print the Dice coefficient:

```python
H = binary
G = ground


dice_GH = (2 * (G * H).sum()/(G.sum() + H.sum()))
print(dice_GH)
```
