**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(img8):

    total = img8.size
    bestT = 0
    min_var = float('inf')
    for T in range(0,256):
        Hg = img8[img8 < T]
        Vg=  img8[img8 > T]
        if Vg.size == 0 or Hg.size == 0:
            continue
       
                
        w1 = Hg.size / total
        w2 = Vg.size / total
        current_variance = w1 * np.var(Hg) + w2 * np.var(Vg)
        if current_variance <= min_var:
            min_var = current_variance
            bestT = T


    
    return bestT
    
                
            

```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

```python
def compute_dice(bin1, bin2):
    threshold = otsu_thresholding(bin2)
    bin1 = bin1.astype(bool)
    bin2 = bin2.astype(bool)
    
    dice_G = (2 * (bin1* bin2).sum()/(bin1.sum() + bin2.sum()))
    
    
    return dice_G
```
